```python
import pandas as pd

data ={
    "Date": pd.date_range("2025-01-01", periods=10, freq='D'),
    "Sales": [100,129,150,130,160,180,200,170,190,220]
}
df=pd.DataFrame(data)
df.set_index("Date" , inplace=True)
print(df)
```

```python
monthly_sales=df["Sales"].resample("ME").sum()
print(monthly_sales)
```

```python
df["Sales"].resample("ME").mean() # average
```

```python
df["Sales"].resample("ME").min() #Minimum
```

```python
df["Sales"].resample("ME").max()
```
